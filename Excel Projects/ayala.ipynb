```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns

```

#### Data Exploration & Cleaning

```python
df = pd.read_excel('ayala1_data.xlsx')
df
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df[['Company', 'Description','Categorization or Classification', 'Vendor', 'Supplier Code/Name', 'Amount', 'Currency', 'Major Category', 'Minor Category']]

df
```

```python
df.drop_duplicates()
```

#### Data Visualization

```python

plt.figure(figsize= (8, 6))
sns.countplot( y = 'Categorization or Classification',
              data = df,
              order = df['Categorization or Classification'].value_counts().nlargest().index,
              color = 'blue'
)
plt.title('Ayala Category Distribution')
```

```python
plt.figure(figsize= (8, 6))
sns.countplot( y = 'Supplier Code/Name',
              data = df,
              order = df['Supplier Code/Name'].value_counts().nlargest(20).index,
              color = 'green'
)
plt.title('Ayala Supplier Distribution')
```

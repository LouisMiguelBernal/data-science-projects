```python
import pandas as pd

# Load the file
df = pd.read_excel("Existing_Categorized.xlsx")  

df.columns = df.columns.str.strip().str.replace(r"\n", " ", regex=True)

desc_cols = [col for col in df.columns if col.startswith("HOSPITAL ITEM DESCRIPTION")]

# Combine all HOSPITAL ITEM DESCRIPTION columns into one column
df["HOSPITAL ITEM DESCRIPTION"] = df[desc_cols] \
    .astype(str) \
    .replace("nan", "", regex=False) \
    .agg(" | ".join, axis=1) \
    .str.strip(" |")

cols_to_drop = [col for col in desc_cols if col != "HOSPITAL ITEM DESCRIPTION"]
df.drop(columns=cols_to_drop, inplace=True)

columns_to_keep_clean = [
    "CATEGORY", "UNIQUE HOSPITAL ITEM CODE", "HOSPITAL ITEM DESCRIPTION", "STATUS",
    "SOURCING CATEGORIZATION - L1", "SOURCING CATEGORIZATION - L2",
    "SEGMENT", "SEGMENT TITLE", "FAMILY", "FAMILY TITLE", "CLASS", "CLASS TITLE",
    "COMMODITY", "COMMODITY TITLE", "SMALL UOM", "BIG UOM", "CON", "NOUN", "TYPE",
    "MANUFACTURER", "MANUFACTURE ITEM CODE", "BRAND/MODEL", "MATERIAL/COMPOSITION",
    "DIMENSION/SIZE (LXBXH)", "DIMENSION/SIZE_ADDITIONAL (LXBXH)", "LENGTH", "LENGTH_ADDITIONAL",
    "SHAPE", "COLOR", "DIAMETER", "ANGLE", "CONFIGURATION", "NO. OF ITEMS", "ITEM_TYPE",
    "LATEX FREE/LATEX", "DISPOSABLE/REUSABLE", "STERILE/NON STERILE", "OTHER ATTRIBUTE",
    "UNSPSC", "UNSPSC DESCRIPTION", "WAVE 1/2 OR 3", "SOURCE DATA FOR MANUFACTURE ITEM CODE",
    "FILE NAME/FILE LINK/WEB LINK FOR MANUFACTURE ITEM CODE", "SOURCE DATA FOR OTHER ATTIBUTES",
    "FILE NAME/FILE LINK/WEB LINK FOR OTHER ATTIBUTES", "FILE NAME/FILE LINK/WEB LINK FOR OTHER ATTIBUTES.1",
    "Done by:", "DATE"
]

columns_present = [col for col in columns_to_keep_clean if col in df.columns]
df = df[columns_present]

# Save the output
df.to_excel("merged_descriptions_output.xlsx", index=False)
```

```python
# Lists

nums = [1, 2, 3, 4, 5]

# Common operations
nums.append(6)          # [1,2,3,4,5,6]
nums.insert(0, 0)       # [0,1,2,3,4,5,6]
nums.pop()              # removes last
nums.remove(3)          # removes first occurrence
nums.sort()             # in-place sort
sorted(nums)            # returns new sorted list
nums.reverse()
len(nums)

# Slicing
nums[1:4]     # [2,3,4]
nums[::-1]    # reversed
nums[::2]     # every other element

# List comprehension ⭐
squares = [x**2 for x in range(10)]
evens = [x for x in range(20) if x % 2 == 0]
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler

import tensorflow as tf
from tensorflow import keras
```

#### Data Overview & Cleaning

```python
df = pd.read_csv('telco.csv')
df.columns = df.columns.str.strip()
df.head()
```

```python
df['Offer'].fillna('None', inplace=True)
df['Internet Type'].fillna('Unknown', inplace=True)
df['Churn Category'].fillna('Unknown', inplace=True)
df['Churn Reason'].fillna('Unknown', inplace=True)
```

```python
df['Churn Label'] = df['Churn Label'].replace({'No': 0, 'Yes': 1})
```

#### Data Preprocessing

```python
# convert caterogrical data into numerical
for col in df.select_dtypes(include='object'):
    df[col] = LabelEncoder().fit_transform(df[col])
```

```python
# split data
x = df.drop(['Churn Label'], axis=1).values
y = df['Churn Label'].values

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=45, shuffle=True)
```

```python
# normalize features

scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test) 
```

```python
# build neural network
model = tf.keras.Sequential([
    tf.keras.layers.Dense(16, activation='relu', input_shape=(x_train.shape[1],)),
    tf.keras.layers.Dense(1, activation='sigmoid')
])
```

```python
# compile & train
model.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
model.fit(x_train, y_train, epochs = 10, batch_size = 32, validation_data = (x_test, y_test))
```

```python
# evaluate model
print('Model Accuracy', model.evaluate(x_test, y_test)[1])
```

```python
from gtts import gTTS
import os

output_folder = "road_marking_audio"
os.makedirs(output_folder, exist_ok=True)

road_markings = {
    "Speed_Limit_60.mp3": "A speed limit of 60 kilometers per hour is the maximum safe speed allowed on this road section.",
    "Speed_Limit_80.mp3": "A speed limit of 80 kilometers per hour is the maximum safe speed allowed on this road section."
}

for filename, text in road_markings.items():
    tts = gTTS(text=text, lang="en")
    tts.save(os.path.join(output_folder, filename))

print("MP3 files created.")
```
